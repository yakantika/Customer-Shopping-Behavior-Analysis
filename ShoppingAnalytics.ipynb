```python
# Q1. Total revenue by male vs female customers
query = """
SELECT gender,
       SUM(puchase_amount) AS total_revenue
FROM customer
GROUP BY gender;
"""

result = pd.read_sql(query, engine)
result
```

```python
# Q2. Customers who used a discount but spent more than average
query = """
SELECT *
FROM customer
WHERE discount_applied = 'Yes'
  AND puchase_amount >
      (SELECT AVG(puchase_amount)
       FROM customer);
"""

result = pd.read_sql(query, engine)
result
```

```python
# Q3. Top 5 products with highest average review rating
query = """
SELECT item_purchased,
       CAST(AVG(review_rating) AS NUMERIC(10,2)) AS avg_rating
FROM customer
GROUP BY item_purchased
ORDER BY avg_rating DESC
LIMIT 5;
"""

result = pd.read_sql(query, engine)
result
```

```python
# Compare the average purchase amount between Standard and Express Shipping
query = """
SELECT shipping_type,
       ROUND(AVG(puchase_amount)::numeric, 2) AS avg_purchase_amount
FROM customer
WHERE shipping_type IN ('Standard', 'Express')
GROUP BY shipping_type;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
# Do subscribed customers spend more? Compare average spend and total revenue between subscribers and non-subscribers.
query = """
SELECT subscription_status,
       ROUND(AVG(puchase_amount)::numeric, 2) AS avg_spend,
       ROUND(SUM(puchase_amount)::numeric, 2) AS total_revenue,
       COUNT(*) AS total_orders
FROM customer
GROUP BY subscription_status;
"""

result = pd.read_sql(query, engine)
print(result)
```

```python
query = """
SELECT customer_segment,
       COUNT(*) AS customer_count
FROM (
    SELECT CASE
        WHEN previous_purchases <= 10 THEN 'New'
        WHEN previous_purchases <= 30 THEN 'Returning'
        ELSE 'Loyal'
    END AS customer_segment
    FROM customer
) AS segments
GROUP BY customer_segment
ORDER BY customer_count DESC;
"""

result = pd.read_sql(query, engine)

result
```

```python
# Q. Top 3 most purchased products within each category
query = """
WITH product_sales AS (
    SELECT category,
           item_purchased,
           COUNT(*) AS purchase_count,
           ROW_NUMBER() OVER (
               PARTITION BY category
               ORDER BY COUNT(*) DESC
           ) AS rank_num
    FROM customer
    GROUP BY category, item_purchased
)

SELECT category,
       item_purchased,
       purchase_count
FROM product_sales
WHERE rank_num <= 3
ORDER BY category, purchase_count DESC;
"""

result = pd.read_sql(query, engine)

result
```
